## Streamlit application setup
Each task writes a Python script and launches a real Streamlit server. The preview is interactive. Only one server runs at a time. Colab uses its authenticated kernel port proxy; if the inline preview is blocked, open the displayed link.

```python
import sys, subprocess, time, urllib.request
from pathlib import Path
from IPython.display import display, IFrame, HTML
server = globals().get("server")
PORT = 8501

def stop_app():
    global server
    if server is not None and server.poll() is None:
        server.terminate()
        try:
            server.wait(timeout=5)
        except subprocess.TimeoutExpired:
            server.kill()
            server.wait()

def run_app(filename):
    global server
    stop_app()
    with open("streamlit_workshop.log", "w") as log:
        server = subprocess.Popen([sys.executable, "-m", "streamlit", "run", filename,
            "--server.port", str(PORT), "--server.headless", "true"], stdout=log, stderr=subprocess.STDOUT)
    for _ in range(60):
        try:
            with urllib.request.urlopen(f"http://127.0.0.1:{PORT}/_stcore/health", timeout=1) as response:
                if response.status == 200:
                    break
        except OSError:
            time.sleep(0.5)
    else:
        raise RuntimeError(Path("streamlit_workshop.log").read_text())
    from google.colab import output
    url = output.eval_js(f"google.colab.kernel.proxyPort({PORT})")
    display(HTML(f'<a href="{url}" target="_blank">Open Streamlit application</a>'))
    display(IFrame(url, width="100%", height=650))
```
